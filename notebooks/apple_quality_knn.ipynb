```python
import pandas as pd
import numpy as np
import sklearn as sklearn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
from scipy import stats
```

```python
df= pd.read_csv('data/apple_problem2.csv')
```

```python
df.head()
```

```python
df['Quality'].unique()
```

```python
# Replace invalid entries with NaN
df.replace(['invalid', 'error', 'NaN', '-9999', '9999'], np.nan, inplace=True)
```

```python
# Convert features to numeric
independant_val = ['Size', 'Weight', 'Sweetness', 'Crunchiness', 'Juiciness', 'Ripeness', 'Acidity']
df[independant_val] = df[independant_val].apply(pd.to_numeric, errors='coerce')
```

```python
# Fill NaN values with the median of each column
for col in independant_val:
    df.fillna(df[col].median(), inplace=True)
```

```python
# Standardize 'Quality' column
df['Quality'] = df['Quality'].str.lower()
```

```python
df['Quality'] = df['Quality'].map({'good': 1, 'bad': 0})
```

```python
df
```

```python
df['Quality'].unique()
```

```python
df
```

```python
# Drop rows with NaN values in 'Quality' column
df = df.dropna(subset=['Quality'])
```

```python
df
```

```python
# Remove outliers using Z-score
z_scores = np.abs(stats.zscore(df[independant_val]))
outliers = df[(z_scores>3).any(axis=1)]
df = df[(z_scores < 3).all(axis=1)]
print(outliers)
```

```python
df['Quality'].unique()
```

```python
df
```

```python
# Split the data into training and test sets
X = df[independant_val]
Y = df['Quality']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Normalize the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Train the KNN model
knn = KNeighborsClassifier(n_neighbors=2) #zID is 5511520
```

```python
knn.fit(X_train, Y_train)
```

```python
# Make predictions and evaluate the model
Y_prediction = knn.predict(X_test)
```

```python
print(confusion_matrix(Y_test, Y_prediction))
print(classification_report(Y_test, Y_prediction))
```
